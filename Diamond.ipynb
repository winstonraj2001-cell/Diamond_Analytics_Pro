```python
# ============================================================
# STEP 24.1: SELECT NUMERICAL FEATURES FOR OUTLIER ANALYSIS
# ============================================================

# Select the numerical columns that we want to analyze
# Price is included because it is our target variable
outlier_columns = ["carat", "depth", "table", "price", "x", "y", "z"]

# Display the selected columns
print("Columns selected for outlier analysis:")
print(outlier_columns)
```

```python
# ============================================================
# STEP 24.2: CREATE BOXPLOTS FOR NUMERICAL FEATURES
# ============================================================

# Import matplotlib
import matplotlib.pyplot as plt

# Create the figure size
plt.figure(figsize=(15, 10))

# Loop through each numerical feature
for i, column in enumerate(outlier_columns, 1):

    # Create a subplot
    plt.subplot(3, 3, i)

    # Create a boxplot
    plt.boxplot(df[column].dropna())

    # Add column name as title
    plt.title(column)

    # Add y-axis label
    plt.ylabel("Value")

# Adjust spacing between plots
plt.tight_layout()

# Display the plots
plt.show()
```

```python
# ============================================================
# STEP 25.1: CALCULATE IQR VALUES
# ============================================================

# Create an empty dictionary to store IQR information
iqr_results = {}

# Loop through every numerical feature
for column in outlier_columns:

    # Calculate the first quartile
    Q1 = df[column].quantile(0.25)

    # Calculate the third quartile
    Q3 = df[column].quantile(0.75)

    # Calculate the Interquartile Range
    IQR = Q3 - Q1

    # Calculate lower boundary
    lower_bound = Q1 - (1.5 * IQR)

    # Calculate upper boundary
    upper_bound = Q3 + (1.5 * IQR)

    # Store the results
    iqr_results[column] = {
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound
    }

# Convert results into a DataFrame
iqr_table = pd.DataFrame(iqr_results).T

# Display the IQR table
iqr_table
```

```python
# ============================================================
# STEP 26: COUNT OUTLIERS USING IQR METHOD
# ============================================================

# Create an empty dictionary to store outlier counts
outlier_counts = {}

# Loop through every numerical feature
for column in outlier_columns:

    # Calculate Q1
    Q1 = df[column].quantile(0.25)

    # Calculate Q3
    Q3 = df[column].quantile(0.75)

    # Calculate IQR
    IQR = Q3 - Q1

    # Calculate lower boundary
    lower_bound = Q1 - (1.5 * IQR)

    # Calculate upper boundary
    upper_bound = Q3 + (1.5 * IQR)

    # Find values below lower boundary
    lower_outliers = (df[column] < lower_bound).sum()

    # Find values above upper boundary
    upper_outliers = (df[column] > upper_bound).sum()

    # Calculate total outliers
    total_outliers = lower_outliers + upper_outliers

    # Store results
    outlier_counts[column] = {
        "Lower Outliers": lower_outliers,
        "Upper Outliers": upper_outliers,
        "Total Outliers": total_outliers
    }

# Convert results into a DataFrame
outlier_table = pd.DataFrame(outlier_counts).T

# Display the outlier table
outlier_table
```

```python
# ============================================================
# STEP 27: CALCULATE OUTLIER PERCENTAGE
# ============================================================

# Create a copy of the outlier table
outlier_summary = outlier_table.copy()

# Calculate the percentage of outliers
outlier_summary["Outlier Percentage"] = (
    outlier_summary["Total Outliers"] / len(df)
) * 100

# Display the summary
outlier_summary
```

```python
# ============================================================
# STEP 28: CHECK SKEWNESS
# ============================================================

# Calculate skewness for numerical features
skewness_values = df[outlier_columns].skew()

# Create a DataFrame for easy viewing
skewness_table = pd.DataFrame({
    "Skewness": skewness_values
})

# Display skewness values
skewness_table
```

```python
# ============================================================
# STEP 29: CLASSIFY SKEWNESS
# ============================================================

# Create a function to classify skewness
def classify_skewness(value):

    # Approximately symmetric
    if -0.5 <= value <= 0.5:
        return "Approximately Symmetric"

    # Moderate skewness
    elif -1 <= value < -0.5 or 0.5 < value <= 1:
        return "Moderately Skewed"

    # High skewness
    else:
        return "Highly Skewed"


# Apply the classification function
skewness_table["Interpretation"] = (
    skewness_table["Skewness"].apply(classify_skewness)
)

# Display the result
skewness_table
```

```python
# ============================================================
# STEP 31: CREATE LOG-TRANSFORMED FEATURES
# ============================================================

# Create a copy of the original DataFrame
df_transformed = df.copy()

# Select positive and highly skewed features
log_columns = ["carat", "price", "x", "y", "z"]

# Apply log1p transformation
# log1p means log(1 + value), which safely handles zero values
for column in log_columns:

    # Create a new transformed column
    df_transformed[f"log_{column}"] = np.log1p(
        df_transformed[column]
    )

# Display the transformed columns
df_transformed[
    [f"log_{column}" for column in log_columns]
].head()
```

```python
# ============================================================
# STEP 32: COMPARE SKEWNESS BEFORE AND AFTER TRANSFORMATION
# ============================================================

# Create an empty list for comparison results
skewness_comparison = []

# Loop through selected columns
for column in log_columns:

    # Calculate original skewness
    original_skewness = df[column].skew()

    # Calculate transformed skewness
    transformed_skewness = df_transformed[f"log_{column}"].skew()

    # Store results
    skewness_comparison.append({
        "Feature": column,
        "Original Skewness": original_skewness,
        "Log Transformed Skewness": transformed_skewness
    })

# Convert list to DataFrame
skewness_comparison_df = pd.DataFrame(skewness_comparison)

# Display comparison
skewness_comparison_df
```

```python
# ============================================================
# STEP 61.8: PREPARE DATA FOR RANDOM FOREST FEATURE IMPORTANCE
# ============================================================

# Import OneHotEncoder
from sklearn.preprocessing import OneHotEncoder

# Import ColumnTransformer
from sklearn.compose import ColumnTransformer

# Define categorical features
categorical_features = [
    "cut",
    "color",
    "clarity"
]

# Define numerical features
numerical_features = [
    "carat",
    "depth",
    "table",
    "x",
    "y",
    "z",
    "volume",
    "dimension_ratio"
]

# Create preprocessing transformer
feature_preprocessor = ColumnTransformer(
    transformers=[
        (
            "categorical",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            ),
            categorical_features
        ),
        (
            "numerical",
            "passthrough",
            numerical_features
        )
    ]
)

# Transform the input features
X_encoded = feature_preprocessor.fit_transform(
    df[categorical_features + numerical_features]
)

# Display encoded data shape
print("Encoded feature shape:", X_encoded.shape)
```

```python
# ============================================================
# STEP 63: TRAIN-TEST SPLIT
# ============================================================

# Import train_test_split from scikit-learn
from sklearn.model_selection import train_test_split

# Split the data into training and testing sets
# 80% of the data is used for training
# 20% of the data is used for testing
# random_state=42 makes the split reproducible
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

# Display the shape of training features
print("X_train shape:", X_train.shape)

# Display the shape of testing features
print("X_test shape:", X_test.shape)

# Display the shape of training target
print("y_train shape:", y_train.shape)

# Display the shape of testing target
print("y_test shape:", y_test.shape)
```

```python
# ============================================================
# STEP 64: DEFINE NUMERICAL AND CATEGORICAL FEATURES
# ============================================================

# Define numerical features
numerical_features = [
    "carat",
    "depth",
    "table",
    "x",
    "y",
    "z",
    "volume",
    "dimension_ratio"
]

# Define categorical features
categorical_features = [
    "cut",
    "color",
    "clarity"
]

# Display numerical features
print("Numerical features:")
print(numerical_features)

# Display categorical features
print("\nCategorical features:")
print(categorical_features)
```

```python
# ============================================================
# STEP 65: ONE-HOT ENCODING
# ============================================================

# Import OneHotEncoder
from sklearn.preprocessing import OneHotEncoder

# Create the One-Hot Encoder
# handle_unknown="ignore" prevents errors if an unseen category
# appears in the test data
encoder = OneHotEncoder(
    handle_unknown="ignore",
    sparse_output=False
)

# Fit the encoder ONLY on the training categorical data
# This prevents test-data information from leaking into training
encoder.fit(X_train[categorical_features])

# Transform training categorical features
X_train_categorical = encoder.transform(
    X_train[categorical_features]
)

# Transform testing categorical features
X_test_categorical = encoder.transform(
    X_test[categorical_features]
)

# Display the encoded shapes
print("Encoded training categorical shape:",
      X_train_categorical.shape)

print("Encoded testing categorical shape:",
      X_test_categorical.shape)
```

```python
# ============================================================
# STEP 66: GET ENCODED FEATURE NAMES
# ============================================================

# Get the names generated by One-Hot Encoding
encoded_feature_names = encoder.get_feature_names_out(
    categorical_features
)

# Display the encoded feature names
print("Encoded categorical features:")

for feature in encoded_feature_names:
    print(feature)
```

```python
# ============================================================
# STEP 67: PREPARE NUMERICAL FEATURES
# ============================================================

# Select numerical columns from training data
X_train_numerical = X_train[numerical_features].copy()

# Select numerical columns from testing data
X_test_numerical = X_test[numerical_features].copy()

# Display training numerical shape
print("Training numerical shape:",
      X_train_numerical.shape)

# Display testing numerical shape
print("Testing numerical shape:",
      X_test_numerical.shape)
```

```python
# ============================================================
# STEP 68: COMBINE NUMERICAL AND ENCODED FEATURES
# ============================================================

# Convert numerical training data into NumPy array
X_train_numerical_array = X_train_numerical.values

# Convert numerical testing data into NumPy array
X_test_numerical_array = X_test_numerical.values

# Combine numerical and categorical features
X_train_encoded = np.hstack([
    X_train_numerical_array,
    X_train_categorical
])

# Combine numerical and categorical features for testing
X_test_encoded = np.hstack([
    X_test_numerical_array,
    X_test_categorical
])

# Display final encoded shapes
print("Final X_train encoded shape:",
      X_train_encoded.shape)

print("Final X_test encoded shape:",
      X_test_encoded.shape)
```

```python
# ============================================================
# STEP 71: SCALE FEATURES
# ============================================================

# Import StandardScaler
from sklearn.preprocessing import StandardScaler

# Create the scaler
scaler = StandardScaler()

# Fit the scaler ONLY on training data
# This prevents information from the test set leaking into training
scaler.fit(X_train_encoded)

# Transform training data
X_train_scaled = scaler.transform(
    X_train_encoded
)

# Transform testing data
X_test_scaled = scaler.transform(
    X_test_encoded
)

# Display the shapes
print("Scaled X_train shape:",
      X_train_scaled.shape)

print("Scaled X_test shape:",
      X_test_scaled.shape)
```

```python
# ============================================================
# STEP 79.2: SET RANDOM SEEDS
# ============================================================

# Set NumPy random seed
np.random.seed(42)

# Set TensorFlow random seed
tf.random.set_seed(42)

# Display confirmation
print("Random seeds set successfully!")
```

```python
# ============================================================
# STEP 79.3: CREATE ANN MODEL
# ============================================================

# Get the number of input features
input_features = X_train_scaled.shape[1]

# Create a Sequential neural network
ann_model = Sequential()

# First hidden layer
# 128 neurons with ReLU activation
ann_model.add(
    Dense(
        128,
        activation="relu",
        input_shape=(input_features,)
    )
)

# Dropout helps reduce overfitting
ann_model.add(
    Dropout(0.2)
)

# Second hidden layer
# 64 neurons with ReLU activation
ann_model.add(
    Dense(
        64,
        activation="relu"
    )
)

# Third hidden layer
# 32 neurons with ReLU activation
ann_model.add(
    Dense(
        32,
        activation="relu"
    )
)

# Output layer
# One neuron because this is a regression problem
ann_model.add(
    Dense(
        1,
        activation="linear"
    )
)

# Display the ANN architecture
ann_model.summary()
```

```python
# ============================================================
# STEP 79.4: COMPILE ANN MODEL
# ============================================================

# Compile the ANN model
# Adam is used as the optimizer
# MSE is used as the loss function
# MAE is included as an additional metric
ann_model.compile(
    optimizer="adam",
    loss="mse",
    metrics=["mae"]
)

# Display confirmation
print("ANN model compiled successfully!")
```

```python
# ============================================================
# STEP 79.5: TRAIN ANN MODEL
# ============================================================

# Import EarlyStopping
from tensorflow.keras.callbacks import EarlyStopping

# Create early stopping callback
# Training stops when validation loss stops improving
early_stopping = EarlyStopping(
    monitor="val_loss",
    patience=10,
    restore_best_weights=True
)

# Train the ANN
ann_history = ann_model.fit(
    X_train_scaled,
    y_train,
    validation_split=0.20,
    epochs=100,
    batch_size=128,
    callbacks=[early_stopping],
    verbose=1
)

# Display confirmation
print("ANN training completed!")
```

```python
# ============================================================
# STEP 79.6: PLOT ANN TRAINING AND VALIDATION LOSS
# ============================================================

# Create the figure
plt.figure(figsize=(10, 6))

# Plot training loss
plt.plot(
    ann_history.history["loss"],
    label="Training Loss"
)

# Plot validation loss
plt.plot(
    ann_history.history["val_loss"],
    label="Validation Loss"
)

# Add title
plt.title("ANN Training and Validation Loss")

# Add x-axis label
plt.xlabel("Epoch")

# Add y-axis label
plt.ylabel("MSE Loss")

# Add legend
plt.legend()

# Display plot
plt.show()
```

```python
# ============================================================
# STEP 86: SELECT FEATURES FOR K-MEANS CLUSTERING
# ============================================================

# Select numerical features that describe diamond characteristics
# Price is intentionally NOT included
clustering_numerical_features = [
    "carat",
    "depth",
    "table",
    "x",
    "y",
    "z",
    "volume",
    "dimension_ratio"
]

# Select categorical quality features
clustering_categorical_features = [
    "cut",
    "color",
    "clarity"
]

# Combine all clustering features
clustering_features = (
    clustering_numerical_features +
    clustering_categorical_features
)

# Create the clustering input dataset
X_cluster = clustering_df[
    clustering_features
].copy()

# Display selected features
print("Features used for K-Means:")
print(clustering_features)

# Display shape
print("\nClustering input shape:")
print(X_cluster.shape)
```

```python
# ============================================================
# STEP 87: CHECK MISSING VALUES IN CLUSTERING DATA
# ============================================================

# Check the number of missing values in each clustering feature
missing_values = X_cluster.isnull().sum()

# Display missing values
print("Missing values in clustering features:")
print(missing_values)

# Calculate the total number of missing values
total_missing = missing_values.sum()

# Display the total
print("\nTotal missing values:", total_missing)

# Check whether any missing values exist
if total_missing == 0:

    # Display success message
    print("\n✅ No missing values found!")
    print("Clustering data is ready for encoding.")

else:

    # Display warning message
    print("\n⚠️ Missing values found!")
    print("We need to handle them before K-Means.")
```

```python
# ============================================================
# STEP 88: ENCODE CATEGORICAL FEATURES FOR K-MEANS
# ============================================================

# Import OneHotEncoder
from sklearn.preprocessing import OneHotEncoder

# Create a new OneHotEncoder for clustering
# handle_unknown="ignore" prevents errors if an unseen category appears
cluster_encoder = OneHotEncoder(
    handle_unknown="ignore",
    sparse_output=False
)

# Encode the categorical clustering features
encoded_cluster_categories = cluster_encoder.fit_transform(
    X_cluster[clustering_categorical_features]
)

# Convert encoded values into a DataFrame
encoded_cluster_df = pd.DataFrame(
    encoded_cluster_categories,
    columns=cluster_encoder.get_feature_names_out(
        clustering_categorical_features
    ),
    index=X_cluster.index
)

# Select the numerical clustering features
cluster_numerical_df = X_cluster[
    clustering_numerical_features
].copy()

# Combine numerical and encoded categorical features
X_cluster_encoded = pd.concat(
    [
        cluster_numerical_df,
        encoded_cluster_df
    ],
    axis=1
)

# Display the shape
print("Encoded clustering data shape:")
print(X_cluster_encoded.shape)

# Display the first five rows
print("\nFirst 5 rows of encoded clustering data:")

display(
    X_cluster_encoded.head()
)
```

```python
# ============================================================
# STEP 89: SCALE K-MEANS CLUSTERING FEATURES
# ============================================================

# Import StandardScaler
from sklearn.preprocessing import StandardScaler

# Create a scaler specifically for clustering
cluster_scaler = StandardScaler()

# Scale all encoded clustering features
X_cluster_scaled = cluster_scaler.fit_transform(
    X_cluster_encoded
)

# Convert the scaled array back into a DataFrame
X_cluster_scaled_df = pd.DataFrame(
    X_cluster_scaled,
    columns=X_cluster_encoded.columns,
    index=X_cluster_encoded.index
)

# Display the shape
print("Scaled clustering data shape:")
print(X_cluster_scaled_df.shape)

# Calculate the mean of the scaled data
cluster_mean = X_cluster_scaled_df.mean().mean()

# Calculate the standard deviation of the scaled data
cluster_std = X_cluster_scaled_df.std().mean()

# Display results
print("\nAverage mean after scaling:", cluster_mean)
print("Average standard deviation after scaling:", cluster_std)

# Display first five rows
print("\nFirst 5 rows of scaled clustering data:")

display(
    X_cluster_scaled_df.head()
)
```

```python
# ============================================================
# STEP 90: ELBOW METHOD FOR OPTIMAL NUMBER OF CLUSTERS
# ============================================================

# Import KMeans
from sklearn.cluster import KMeans

# Create an empty list to store the Within-Cluster Sum of Squares
inertia_values = []

# Define the range of K values to test
k_values = range(2, 9)

# Test each K value
for k in k_values:

    # Create the K-Means model
    kmeans_model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    # Train K-Means on the scaled clustering data
    kmeans_model.fit(
        X_cluster_scaled
    )

    # Store the inertia value
    inertia_values.append(
        kmeans_model.inertia_
    )

    # Display progress
    print(
        f"K = {k} | Inertia = {kmeans_model.inertia_:,.2f}"
    )

# Create the Elbow plot
plt.figure(figsize=(10, 6))

# Plot K against inertia
plt.plot(
    list(k_values),
    inertia_values,
    marker="o"
)

# Add title
plt.title(
    "Elbow Method for Selecting Optimal K"
)

# Add x-axis label
plt.xlabel(
    "Number of Clusters (K)"
)

# Add y-axis label
plt.ylabel(
    "Inertia"
)

# Add grid
plt.grid(
    True,
    alpha=0.3
)

# Display the plot
plt.show()
```

```python
# ============================================================
# STEP 91: SILHOUETTE SCORE FOR CLUSTER VALIDATION
# ============================================================

# Import silhouette_score
from sklearn.metrics import silhouette_score

# Set the sample size for silhouette calculation
silhouette_sample_size = min(
    5000,
    len(X_cluster_scaled)
)

# Create a reproducible random sample
silhouette_sample = X_cluster_scaled_df.sample(
    n=silhouette_sample_size,
    random_state=42
)

# Create an empty list to store silhouette scores
silhouette_scores = []

# Test each K value
for k in k_values:

    # Create K-Means model
    kmeans_model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    # Fit K-Means on the complete clustering dataset
    kmeans_model.fit(
        X_cluster_scaled
    )

    # Predict cluster labels for the sample
    sample_labels = kmeans_model.predict(
        silhouette_sample
    )

    # Calculate Silhouette Score on the sample
    score = silhouette_score(
        silhouette_sample,
        sample_labels
    )

    # Store the score
    silhouette_scores.append(
        score
    )

    # Display the result
    print(
        f"K = {k} | Silhouette Score = {score:.4f}"
    )

# Create the Silhouette Score plot
plt.figure(figsize=(10, 6))

# Plot K against Silhouette Score
plt.plot(
    list(k_values),
    silhouette_scores,
    marker="o"
)

# Add title
plt.title(
    "Silhouette Score for Different K Values"
)

# Add x-axis label
plt.xlabel(
    "Number of Clusters (K)"
)

# Add y-axis label
plt.ylabel(
    "Silhouette Score"
)

# Add grid
plt.grid(
    True,
    alpha=0.3
)

# Display plot
plt.show()
```

```python
# ============================================================
# STEP 92: COMPARE ELBOW AND SILHOUETTE RESULTS
# ============================================================

# Create a DataFrame containing the evaluation results
cluster_evaluation = pd.DataFrame({
    "K": list(k_values),
    "Inertia": inertia_values,
    "Silhouette_Score": silhouette_scores
})

# Display the evaluation table
print("K-Means Cluster Evaluation:")
display(
    cluster_evaluation
)

# Find the K with the highest Silhouette Score
best_silhouette_k = (
    cluster_evaluation.loc[
        cluster_evaluation[
            "Silhouette_Score"
        ].idxmax(),
        "K"
    ]
)

# Display the best silhouette K
print(
    "\nBest K based on Silhouette Score:",
    int(best_silhouette_k)
)
```

```python
# ============================================================
# STEP 93: TRAIN FINAL K-MEANS CLUSTERING MODEL
# ============================================================

# Get the optimal K from the highest Silhouette Score
optimal_k = int(best_silhouette_k)

# Display the selected K
print("Selected optimal number of clusters:", optimal_k)

# Create the final K-Means model
final_kmeans_model = KMeans(
    n_clusters=optimal_k,
    random_state=42,
    n_init=10
)

# Train the final K-Means model
final_kmeans_model.fit(
    X_cluster_scaled
)

# Generate cluster labels for every diamond
cluster_labels = final_kmeans_model.labels_

# Add cluster labels to a copy of the original data
clustered_df = clustering_df.copy()

# Create the cluster column
clustered_df["cluster"] = cluster_labels

# Display confirmation
print("\n✅ Final K-Means model trained successfully!")

# Display number of observations
print(
    "Number of clustered diamonds:",
    len(clustered_df)
)
```

```python
# ============================================================
# STEP 94: ANALYZE CLUSTER SIZES
# ============================================================

# Count diamonds in each cluster
cluster_sizes = (
    clustered_df["cluster"]
    .value_counts()
    .sort_index()
)

# Display cluster sizes
print("Number of diamonds in each cluster:")
print(cluster_sizes)

# Calculate percentage of diamonds in each cluster
cluster_percentages = (
    clustered_df["cluster"]
    .value_counts(
        normalize=True
    )
    .sort_index()
    * 100
)

# Display percentages
print("\nPercentage of diamonds in each cluster:")

for cluster_id, percentage in cluster_percentages.items():

    print(
        f"Cluster {cluster_id}: "
        f"{percentage:.2f}%"
    )
```

```python
# ============================================================
# STEP 95: CREATE CLUSTER PROFILES
# ============================================================

# Calculate average numerical characteristics for each cluster
cluster_numeric_profile = (
    clustered_df
    .groupby("cluster")[
        [
            "carat",
            "depth",
            "table",
            "x",
            "y",
            "z",
            "volume",
            "dimension_ratio",
            "price"
        ]
    ]
    .mean()
    .round(3)
)

# Display numerical cluster profiles
print("Numerical Cluster Profiles:")
display(
    cluster_numeric_profile
)
```

```python
# ============================================================
# STEP 96: ANALYZE CATEGORICAL CLUSTER CHARACTERISTICS
# ============================================================

# Create an empty list to store categorical profiles
categorical_profiles = []

# Loop through every cluster
for cluster_id in sorted(
    clustered_df["cluster"].unique()
):

    # Select only the current cluster
    cluster_data = clustered_df[
        clustered_df["cluster"] == cluster_id
    ]

    # Find the most common cut
    most_common_cut = (
        cluster_data["cut"]
        .mode()
        .iloc[0]
    )

    # Find the most common color
    most_common_color = (
        cluster_data["color"]
        .mode()
        .iloc[0]
    )

    # Find the most common clarity
    most_common_clarity = (
        cluster_data["clarity"]
        .mode()
        .iloc[0]
    )

    # Add profile information
    categorical_profiles.append(
        {
            "Cluster": cluster_id,
            "Most Common Cut": most_common_cut,
            "Most Common Color": most_common_color,
            "Most Common Clarity": most_common_clarity,
            "Diamond Count": len(cluster_data)
        }
    )

# Convert profiles into a DataFrame
categorical_profile_df = pd.DataFrame(
    categorical_profiles
)

# Display categorical profiles
print("Categorical Cluster Profiles:")

display(
    categorical_profile_df
)
```

```python
# ============================================================
# STEP 97: CREATE COMPLETE CLUSTER PROFILE
# ============================================================

# Calculate average price for each cluster
cluster_price = (
    clustered_df
    .groupby("cluster")["price"]
    .mean()
    .round(2)
    .reset_index()
)

# Rename price column
cluster_price.columns = [
    "Cluster",
    "Average Price"
]

# Calculate average carat for each cluster
cluster_carat = (
    clustered_df
    .groupby("cluster")["carat"]
    .mean()
    .round(3)
    .reset_index()
)

# Rename carat column
cluster_carat.columns = [
    "Cluster",
    "Average Carat"
]

# Calculate average volume
cluster_volume = (
    clustered_df
    .groupby("cluster")["volume"]
    .mean()
    .round(3)
    .reset_index()
)

# Rename volume column
cluster_volume.columns = [
    "Cluster",
    "Average Volume"
]

# Merge cluster information
complete_cluster_profile = (
    categorical_profile_df
    .merge(
        cluster_price,
        on="Cluster"
    )
    .merge(
        cluster_carat,
        on="Cluster"
    )
    .merge(
        cluster_volume,
        on="Cluster"
    )
)

# Sort clusters by average price
complete_cluster_profile = (
    complete_cluster_profile
    .sort_values(
        "Average Price",
        ascending=False
    )
    .reset_index(drop=True)
)

# Display complete profile
print("COMPLETE CLUSTER PROFILE")
print("========================")

display(
    complete_cluster_profile
)
```

```python
# ============================================================
# STEP 98: ASSIGN MEANINGFUL CLUSTER NAMES
# ============================================================

# Create a copy of the profile
named_cluster_profile = (
    complete_cluster_profile.copy()
)

# Create rank based on average price
named_cluster_profile["Price Rank"] = (
    named_cluster_profile[
        "Average Price"
    ]
    .rank(
        ascending=False,
        method="first"
    )
)

# Create an empty dictionary for cluster names
cluster_name_mapping = {}

# Number of clusters
number_of_clusters = len(
    named_cluster_profile
)

# Assign meaningful names based on price ranking
for _, row in named_cluster_profile.iterrows():

    # Get cluster ID
    cluster_id = int(
        row["Cluster"]
    )

    # Get price rank
    price_rank = int(
        row["Price Rank"]
    )

    # Highest-priced cluster
    if price_rank == 1:

        cluster_name_mapping[
            cluster_id
        ] = "Premium Diamonds"

    # Lowest-priced cluster
    elif price_rank == number_of_clusters:

        cluster_name_mapping[
            cluster_id
        ] = "Value Diamonds"

    # Remaining clusters
    else:

        cluster_name_mapping[
            cluster_id
        ] = f"Standard Segment {price_rank - 1}"

# Add meaningful names
named_cluster_profile["Segment Name"] = (
    named_cluster_profile["Cluster"]
    .map(cluster_name_mapping)
)

# Display final named segments
print("MARKET SEGMENTS")

display(
    named_cluster_profile[
        [
            "Cluster",
            "Segment Name",
            "Diamond Count",
            "Average Price",
            "Average Carat",
            "Average Volume",
            "Most Common Cut",
            "Most Common Color",
            "Most Common Clarity"
        ]
    ]
)
```

```python
# ============================================================
# STEP 99: ADD MARKET SEGMENT NAMES TO DATASET
# ============================================================

# Add the cluster ID
clustered_df["cluster_id"] = (
    clustered_df["cluster"]
)

# Add the meaningful segment name
clustered_df["segment_name"] = (
    clustered_df["cluster"]
    .map(cluster_name_mapping)
)

# Display sample records
print("Sample records with market segments:")

display(
    clustered_df[
        [
            "carat",
            "cut",
            "color",
            "clarity",
            "price",
            "cluster_id",
            "segment_name"
        ]
    ].head(10)
)
```

```python
# ============================================================
# STEP 100: PCA FOR K-MEANS CLUSTER VISUALIZATION
# ============================================================

# Import PCA from scikit-learn
from sklearn.decomposition import PCA

# Create PCA with 2 components
# We use 2 components so the clusters can be displayed in 2D
pca = PCA(
    n_components=2,
    random_state=42
)

# Fit PCA on the scaled clustering data
# PCA is used only to reduce dimensions for visualization
X_cluster_pca = pca.fit_transform(
    X_cluster_scaled
)

# Convert PCA results into a DataFrame
pca_df = pd.DataFrame(
    X_cluster_pca,
    columns=[
        "PCA1",
        "PCA2"
    ],
    index=clustered_df.index
)

# Add the cluster ID to the PCA DataFrame
pca_df["cluster_id"] = (
    clustered_df["cluster_id"]
)

# Add the meaningful segment name
pca_df["segment_name"] = (
    clustered_df["segment_name"]
)

# Display PCA data shape
print("PCA data shape:")
print(pca_df.shape)

# Display explained variance ratio
print("\nExplained variance ratio:")

print(
    "PCA1:",
    f"{pca.explained_variance_ratio_[0]:.4f}"
)

print(
    "PCA2:",
    f"{pca.explained_variance_ratio_[1]:.4f}"
)

# Calculate total variance explained by the two components
total_explained_variance = (
    pca.explained_variance_ratio_.sum()
)

# Display total explained variance
print(
    "\nTotal variance explained by PCA1 + PCA2:",
    f"{total_explained_variance:.4f}"
)

# Display first five PCA records
print("\nFirst 5 PCA records:")

display(
    pca_df.head()
)
```

```python
# ============================================================
# STEP 101: 2D K-MEANS CLUSTER VISUALIZATION USING PCA
# ============================================================

# Import Plotly Express
import plotly.express as px

# Create a 2D scatter plot using the PCA results
fig = px.scatter(
    pca_df,
    x="PCA1",
    y="PCA2",
    color="segment_name",
    hover_data=[
        "cluster_id",
        "segment_name"
    ],
    title="Diamond Market Segments - PCA Visualization",
    labels={
        "PCA1": "Principal Component 1",
        "PCA2": "Principal Component 2",
        "segment_name": "Market Segment"
    }
)

# Set chart height
fig.update_layout(
    height=650
)

# Display the interactive chart
fig.show()
```

```python
# ============================================================
# STEP 102: MARKET SEGMENT DISTRIBUTION
# ============================================================

# Import Plotly Express
import plotly.express as px

# Count the number of diamonds in each market segment
segment_distribution = (
    clustered_df["segment_name"]
    .value_counts()
    .reset_index()
)

# Rename the columns
segment_distribution.columns = [
    "Segment",
    "Diamond Count"
]

# Calculate the percentage of diamonds in each segment
segment_distribution["Percentage"] = (
    segment_distribution["Diamond Count"]
    / len(clustered_df)
    * 100
)

# Round the percentage values
segment_distribution["Percentage"] = (
    segment_distribution["Percentage"]
    .round(2)
)

# Display the segment distribution table
print("MARKET SEGMENT DISTRIBUTION")
print("===========================")

display(
    segment_distribution
)

# Create an interactive bar chart
fig = px.bar(
    segment_distribution,
    x="Segment",
    y="Diamond Count",
    text="Diamond Count",
    title="Number of Diamonds in Each Market Segment",
    labels={
        "Segment": "Market Segment",
        "Diamond Count": "Number of Diamonds"
    }
)

# Display the values on top of the bars
fig.update_traces(
    textposition="outside"
)

# Set chart height
fig.update_layout(
    height=500
)

# Display the interactive chart
fig.show()
```

```python
# ============================================================
# STEP 103: BUSINESS INTERPRETATION OF MARKET SEGMENTS
# ============================================================

# Create a business-oriented cluster summary
business_segment_summary = (
    clustered_df
    .groupby("segment_name")
    .agg(
        diamond_count=("carat", "count"),
        average_carat=("carat", "mean"),
        average_price=("price", "mean"),
        average_volume=("volume", "mean"),
        average_depth=("depth", "mean"),
        average_table=("table", "mean")
    )
    .round(2)
    .reset_index()
)

# Calculate the percentage of diamonds in each segment
business_segment_summary["market_share_percent"] = (
    business_segment_summary["diamond_count"]
    / len(clustered_df)
    * 100
)

# Round market share
business_segment_summary["market_share_percent"] = (
    business_segment_summary["market_share_percent"]
    .round(2)
)

# Sort segments from highest average price to lowest
business_segment_summary = (
    business_segment_summary
    .sort_values(
        "average_price",
        ascending=False
    )
    .reset_index(drop=True)
)

# Display the business summary
print("============================================================")
print("          DIAMOND MARKET SEGMENT BUSINESS SUMMARY")
print("============================================================")

display(
    business_segment_summary
)
```

```python
# ============================================================
# STEP 103.1: GENERATE BUSINESS INTERPRETATION
# ============================================================

# Identify the highest-priced segment
highest_price_segment = (
    business_segment_summary
    .iloc[0]
)

# Identify the largest segment by number of diamonds
largest_segment = (
    business_segment_summary
    .sort_values(
        "diamond_count",
        ascending=False
    )
    .iloc[0]
)

# Identify the lowest-priced segment
lowest_price_segment = (
    business_segment_summary
    .sort_values(
        "average_price",
        ascending=True
    )
    .iloc[0]
)

# Display business insights
print("BUSINESS INSIGHTS")
print("=================")

print(
    f"\n💎 Premium Segment:"
    f" {highest_price_segment['segment_name']}"
)

print(
    f"   Average price: "
    f"${highest_price_segment['average_price']:,.2f}"
)

print(
    f"   Average carat: "
    f"{highest_price_segment['average_carat']:.2f}"
)

print(
    f"\n📦 Largest Market Segment:"
    f" {largest_segment['segment_name']}"
)

print(
    f"   Diamonds: "
    f"{int(largest_segment['diamond_count']):,}"
)

print(
    f"   Market share: "
    f"{largest_segment['market_share_percent']:.2f}%"
)

print(
    f"\n💰 Value Segment:"
    f" {lowest_price_segment['segment_name']}"
)

print(
    f"   Average price: "
    f"${lowest_price_segment['average_price']:,.2f}"
)

print(
    f"   Average carat: "
    f"{lowest_price_segment['average_carat']:.2f}"
)
```
